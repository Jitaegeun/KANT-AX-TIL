````python
# * 튜플 , ** 딕셔너리

def total(*nums): # * 사용하여 nums 매개변수 입력하면 튜플로 묶여서 처리됨
    return sum(nums)

total(1, 2) # () 튜플로 묶임
total(1, 2, 3, 4, 5)

def info(**data):
    for key, value in data.items():
        print(key,":", value)

info(이름="Kim", 경력=7)
# 이름 : KIM
# 경력 : 7   
# 딕셔너리로 묶임
````

````python
# 함수 활용
# 함수가 어떤 경우든 결과를 같은 자료형으로 돌려줘야 한다. (중요)

def get_average(*scores): # *튜플
    if len(scores) == 0: #len : 길이
        return 0
    total = sum(scores)
    return total / len(scores)  

print(f"{get_average(80, 90, 100):.2f}")  # :.2f 소수점 둘째자리 까지 출력
print(f"{get_average(75, 60, 88, 92, 100, 45):.2f}")

print(f"{get_average(75,88,95,100,50):.2f}") # Q : print 안써도 출력이 되네. A : 쥬피터 노트북의 기능임 / print 쓰는 버릇을 들여라
# f"" : f-문자열 시작과 끝
# {}  : "여기에는 글자 말고 계산 결과를 넣어라"
# (소괄호): 함수호출 , "따옴표" : 문자열, {중괄호} = 값
````

````python
# 지역 변수와 전역 변수
[1]
message = "저는 전역 변수입니다"

def test_scope():
    message = "저는 지역 변수입니다"
    print(f"[함수 내부 출력] {message}")

test_scope()
print(f"[함수 외부 출력] {message}")
print("-"*30)
[2]
# 전역변수
message = "안녕하세요. Kant몰에 오신것을 환영합니다."

def greeting():
    message = "장난감 코너는 3층에 있습니다."
    print(message) # 함수 안에서 같은 이름이 있으면 -> 지역변수가 우선된다.
    
greeting()

print(message) # 함수 밖에서는 지역변수가 이미 사라졌으니 -> 전역변수
print("-"*30)

[3]
# 전역변수
value = "전역 데이터"

def outer_function():
    value = "지역 변수 1"       # 2. outer 안에 value 생성
    
    def inner_function():  
        value = "지역 변수 2"   # 4. inner 안에 value 생성 
        print(value)           # 5. 출력 : 지역 변수 2
        
    inner_function()           # 3. inner 호출
    print(value)               # 6. 출력 지역 변수 1
    
    
outer_function()               # 1. 시작
print(value) #전역변수          # 7. 출력 : 전역 데이터
````

### LEGB 규칙


내부 데이터 -> 외부 데이터 -> 전역 데이터 -> Built-in




````python
value = "전역 데이터"              # G

def outer_function():
    value = "외부 데이터"          # E
    def inner_function():
        value = "내부 데이터"      # L
        print(value)
    inner_function()

outer_function()
````

````python
# Global 과 UnboundLocalError
# unboundLocalError : 함수 안의 지역변수에 아직 값이 들어가지 않았는데 그 값을 가져다가 쓰려고 할때
counter = 5

def increase_counter():
    global counter    # 전역변수 선언 안하면 UnboundLocalError 발생
    #counter = 0         # counter 값을 지정했음 counter = 0 그러면 이거는 지역변수로 인지 , 전역변수 5랑 다른걸로 봄
    counter += 1

increase_counter()
increase_counter()
print(f"최종 카운트 결과값: {counter}")

def counter_func(current):
    return current + 1
````

````python
# lambda·map·filter

numbers_list = [1, 2, 3, 4, 5]
print(list(map(lambda x: x + 10, numbers_list)))
print(list(filter(lambda x: x % 2 == 0, numbers_list)))   #조건 : # % 2 == 0 2로 나눈 나머지가 0인것
print("-"*30)
m = map(lambda x: x + 10, numbers_list) # 아무 계산하지 않는다. "numbers_list"의 각 값에 +10을 해라 라는 지시서만 m에 저장
print(m)    # map은 계산 결과를 바로 만들지 않고, 하나씩 꺼낼 수 있는 "통로(이터레이터)"만 만든다. 저 값은 객체정보
            # map은 일회용 주문서 같은 개념
print(list(m)) # 처음 호출할 때 통로를 한 번 끝까지 지나가면서 값을 꺼내고, 통로가 비어서 두번째 리스트는 빈 리스트가 된다.
print(list(m)) # 이미 앞에서 통로 안의 값이 모두 사용되어 없어짐. 빈통로라서 [] 가 출력됨.
````

요약 : map은 바로 계산하지 않고, 값을 한 번만 꺼낼 수 있는 객체를 돌려준다 그래서 list()로 한번 담아야 여러 번 쓸 수 있다.

````python
# 정렬 기준과 종합
# 람다의 대표 용도 : sorted · max · min 의 key

students = [("민수", 85), ("영희", 92), ("철수", 78)]
print(sorted(students, key=lambda s: s[1])) #sorted : 정렬된 새 리스트를 만들어 돌려주는 함수
print(max(students, key=lambda s: s[1]))

raw_scores = [2, 3, 4, 5, 6, 7]
squared = list(map(lambda n: n ** 2, raw_scores)) # raw_scores의 값을 제곱하였음. / map(함수,리스트) list(): 인수 하나만 받을 수 있음.
print(list(filter(lambda score: score > 20, squared)))
````
