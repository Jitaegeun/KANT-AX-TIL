````python
# 기본값 매개변수

[1] # 매개변수의 기본값이 있으면 뒤에 배치해라.
def introduce(name,age=20):
    print(f"이름은{name},나이는{age}세 입니다.")
introduce("철수")          # def 함수에 age 기본값20을 설정하면 인수를 생략해도 됨.
introduce("영희",30)
introduce("나영",age = 12) # 키워드 인수 사용

[2] #기본값을 설정할때 리스트[] 나 딕셔너리{} 같은 자료형을 쓰지말고 None을 넣은 뒤 함수를 만들어라
# 파이썬의 유명한 "기본값으로 리스트를 쓰면 안되는 이유"
def add_item(item, basket=[]):
    basket.append(item)
    return basket

print(add_item("사과"))
print(add_item("바나나"))
print(add_item("체리"))

[2-2]
def add_item(item, basket=None):
    if basket is None:
        basket = []            # 호출될 때마다 바구니를 비움
    basket.append(item)
    return basket

print(add_item("사과"))
print(add_item("바나나"))
print(add_item("체리"))
````

````python
# * 튜플 , ** 딕셔너리

def total(*nums): # * 사용하여 nums 매개변수 입력하면 튜플로 묶여서 처리됨
    return sum(nums)

total(1, 2) # () 튜플로 묶임
total(1, 2, 3, 4, 5)

def info(**data):
    for key, value in data.items():
        print(key,":", value)

info(이름="Kim", 경력=7)
# 이름 : KIM
# 경력 : 7   
# 딕셔너리로 묶임
````

````python
# 함수 활용
# 함수가 어떤 경우든 결과를 같은 자료형으로 돌려줘야 한다. (중요)

def get_average(*scores): # *튜플
    if len(scores) == 0: #len : 길이
        return 0
    total = sum(scores)
    return total / len(scores)  

print(f"{get_average(80, 90, 100):.2f}")  # :.2f 소수점 둘째자리 까지 출력
print(f"{get_average(75, 60, 88, 92, 100, 45):.2f}")

print(f"{get_average(75,88,95,100,50):.2f}") # Q : print 안써도 출력이 되네. A : 쥬피터 노트북의 기능임 / print 쓰는 버릇을 들여라
# f"" : f-문자열 시작과 끝
# {}  : "여기에는 글자 말고 계산 결과를 넣어라"
# (소괄호): 함수호출 , "따옴표" : 문자열, {중괄호} = 값
````

````python
# 지역 변수와 전역 변수
[1]
message = "저는 전역 변수입니다"

def test_scope():
    message = "저는 지역 변수입니다"
    print(f"[함수 내부 출력] {message}")

test_scope()
print(f"[함수 외부 출력] {message}")
print("-"*30)
[2]
# 전역변수
message = "안녕하세요. Kant몰에 오신것을 환영합니다."

def greeting():
    message = "장난감 코너는 3층에 있습니다."
    print(message) # 함수 안에서 같은 이름이 있으면 -> 지역변수가 우선된다.
    
greeting()

print(message) # 함수 밖에서는 지역변수가 이미 사라졌으니 -> 전역변수
print("-"*30)

[3]
# 전역변수
value = "전역 데이터"

def outer_function():
    value = "지역 변수 1"       # 2. outer 안에 value 생성
    
    def inner_function():  
        value = "지역 변수 2"   # 4. inner 안에 value 생성 
        print(value)           # 5. 출력 : 지역 변수 2
        
    inner_function()           # 3. inner 호출
    print(value)               # 6. 출력 지역 변수 1
    
    
outer_function()               # 1. 시작
print(value) #전역변수          # 7. 출력 : 전역 데이터
````
