### 2. 오늘 학습한 개념 : 무엇을 배웠는지
docker 프로그램 : 개발자PC , 동료 PC, 테스트 서버, 운영 서버마다 OS 버전, 설치된 라이브러리, 설정이 조금씩 다르더라.
그래서 같은 코드인데 어디서는 구동이 되고 어디서는 안됨.
애플리케이션과 실행에 필요한 모든 것을 하나의 "이미지"로 포장 / 똑같은 "컨테이너"로 실행

images 는 게임 CD 이고 Containers 는 구동하는 게임기야

Redis - 인메모리 Key-Value 저장소

볼륨 - 

Docker Compose - 여러 컨테이너를 파일 하나로




### 3. 이해한 내용 : 내 말로 설명
 - Docker는 개발 서버를 통일하기 위한것 
   프론트엔드 백엔드 그외에도 개발 상황에 따라 환경도 다름
 - images : 닌텐도 칩, 게임 CD 같은거임.
 - Containers : 구동 
 - Docker hub : 전문가들이 패키지로 만들어서 올려둔 홈페이지
