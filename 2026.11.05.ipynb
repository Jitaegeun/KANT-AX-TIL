## 문제 3 — 은행 계좌 입출금 프로그램 (보통)

**목표:** 클래스 메서드에 계좌 동작을 구현하고 `while`과 조건문으로 메뉴 만들기

**요구사항**
1. `BankAccount(owner, balance)`로 소유자와 잔액을 저장합니다.
2. `deposit(amount)`는 **양수만** 입금하고 성공 여부를 알려줍니다.
3. `withdraw(amount)`는 양수이며 잔액 이하일 때만 출금합니다.
4. `show()`는 현재 잔액 문자열을 반환합니다.
5. 메뉴 1 입금 / 2 출금 / 3 잔액 확인 / 4 종료가 반복되게 만듭니다.
6. 잘못된 메뉴나 숫자 입력도 프로그램이 갑자기 종료되지 않게 처리해 보세요. *(도전 항목)*

**예시 흐름**
```text
1. 입금
2. 출금
3. 잔액 확인
4. 종료
메뉴 선택: 1
입금 금액: 5000
입금 완료
...
메뉴 선택: 3
현재 잔액: 15000원
...
메뉴 선택: 4
프로그램 종료
```

**힌트:** `while True`, `if / elif / else`, `break`, `int(input(...))`. 입력 검증은 `try / except ValueError`를 활용할 수 있습니다.

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount   # self.balance 로 수정, self.amount로 적으면 안됨. 
        print (f"{amount} 원 입금 완료하였습니다.")  # self.amount 가 아님, amount로 수정
        print (f"총 액 : {self.balance} 원 입니다.") 
        
        
    def withdraw(self, amount):
        # self.amount -= amount
        if amount <= self.balance: # 출금 가능 여부만 판단함.
            self.balance -= amount   # 이렇게 해야 출금하면 잔액에서 빠져나감  / 출금 여부 판단하고 돈을 뺴자
            print ("출금 완료.") # 출력값 수정,퀄리티
            print (f"현재 잔액 : {self.balance}원")
        # 잔액이 없을때 못뽑게 하기
        else:     
            print ("잔액이 부족합니다.")
    # 잔액 확인    
    def show(self):
        return(f"{self.balance}원")
print("종료")    

# account = BankAccount("지태근", 0)   #객체 쓸 줄 모름 / 객체는 while 문 앞에

#  이름을 딕셔너리로 하고 이전에 입력한 사람이면 기존 계좌를 꺼내고, 처음 입력한 사람이면 초기잔액 0으로 새 계좌 생성, 저장
accounts = {} #딕셔너리

owner = input("계좌 소유자 이름 : ") # 이름 -> 객체로 저장하는 곳

if owner in accounts: # 이전에 입력한 사람
    account = accounts[owner] #기존 계좌로 넘어감
else: #처음 입력한 사람
    account = BankAccount(owner, 0) #초기 잔액 0 으로 새 계좌 생성 
    accounts[owner] = account #저장

while True:
    print("1. 입금" )
    print("2. 출금" )
    print("3. 잔액 확인" )
    print("4. 종료" )
    
    menu = input("메뉴 선택 : ")
    
    if menu == "1":
        amount = int(input("입금 금액 : "))
        account.deposit(amount)
    
    elif menu == "2":
        amount = int(input("출금 금액 : "))
        account.withdraw(amount)
    
    elif menu == "3":
        print(f"현재 잔액 :{account.show()}원")
        
    elif menu == "4":
        print("종료")
        break
    
    else:
        print ("아직 구현하지 않은 메뉴 입니다.")
        # return 은 여기 쓰는 개념이 아닌듯
````
