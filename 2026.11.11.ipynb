### 6. 코드 이해 

````python
class Product:    # 왜 Product 로 클래스를 만들었어? 
    def __init__(self, name, price, stock): # 이름, 가격, 개수
        self.name = name   # 인스턴스 변수
        self.price = price
        self.stock = stock
    
    def __str__(self):   # str 출력값을 customer 가 보기좋게끔.
        return f"{self.id} : {self.name:<10} {self.price:>8,} {self.stock}" # :<10 10칸 확보 왼쪽 정렬 / :>8, 8칸 확보하고 오른쪽 정렬 , 천 단위 쉼표 
    
# ----------------------------------------
# 테스트 코드
# ----------------------------------------
earphone = Product("무선이어폰", 80_000, 3)
tumbler = Product("텀블러", 15_000, 8)

print(earphone.name, earphone.price, earphone.stock)
print(tumbler.name, tumbler.price, tumbler.stock)

````
