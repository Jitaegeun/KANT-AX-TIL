### 3. 이해한 내용 : 내 말로 설명


```python
# 절차지향 : 데이터와 함수가 따로
user_name = "스파르타"  # 데이터
print (f"{user_name}")  # 함수
# print_user_name(user_name)

#객체지향 : 상태와 행동을 하나로 묶음.
class User:    #Class 사용
    def __init__(self, name): 
        self.name = name
    def show(self):
        print(f"회원 이름: {self.name}")

user_instance = User("스파르타") #인스턴스
user_instance.show()
# 리스트도 객체
```
