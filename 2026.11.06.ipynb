#### 문제 — 학생 성적 관리 시스템 (응용)

**목표:** 클래스, `for`, `while`, `lambda`, `sorted()`를 결합해 성적 순위를 출력하기

**요구사항**
1. `Student` 인스턴스 4개를 리스트에 저장합니다.
2. 메뉴 1은 `for`로 전체 학생 정보를 출력합니다.
3. 메뉴 2는 `sorted(students, key=..., reverse=True)`로 점수가 높은 순서로 정렬해 순위를 출력합니다.
4. `key` 부분에 `lambda student: student.score`를 활용합니다.
5. 메뉴 3은 프로그램을 종료합니다.

**메뉴 2 예상 결과**
```text
=== 학생 점수 순위 ===
1등 : 영희 - 95점
2등 : 민수 - 90점
3등 : 지수 - 85점
4등 : 철수 - 75점
```

**힌트:** `enumerate(정렬된_리스트, start=1)`를 사용하면 1등부터 번호를 매길 수 있습니다. `lambda`는 `sorted`가 비교할 점수 값을 알려주는 함수입니다.

````python
class student:    # class 생성
    def __init__(self, name, score):
        self.name = name
        self.score = score
    
students = [    # = 빠짐
    student("영희",95),
    student("민수",90),
    student("지수",85),
    student("철수",75)
]
from IPython.display import clear_output    # 출력문이 계속 쌓이는것을 방지하기 위함

while True:
    clear_output(wait=True)  # 출력문이 계속 쌓이는것을 방지하기 위함
    
    print("메뉴 1 전체 학생 정보")
    print("메뉴 2 성적 순위")
    print("메뉴 3 종료")
    
    menu = input("메뉴 1~3 중에 선택하시오.")

    if menu == "1":
        print("전체 학생 정보")
        for student in students:
            print(f"{student.name} - {student.score}") #student.show 로 변경 후 show 철회
    elif menu == "2":
        print("성적 순위")   # 전혀 감이 안잡힘
        ranked = sorted(    #정렬 
            students,
            key=lambda student: student.score,   #lambda 로 이름과 점수를 출력?  #한줄로쓴다............. 원래 키값을 연결해야하는데 한줄로쓰려고..
            reverse=True  #내림차순6544
        )
        for rank, student in enumerate(ranked, start=1):   #리스트에 번호를 매겨주는것
            print(f"{rank}등 : {student.name} - {student.score}점")

    elif menu == "3":
        print("프로그램을 종료합니다")
        break
    else:
        print("선택하실 수 없습니다.")

    input("\n결과를 확인했다면 Enter를 누르세요.")  # 출력문이 계속 쌓이는것을 방지하기 위함
````
