#### 놀이공원 입장 요금 계산기
- 나이가 3살 미만이면 무료 입장
- 3살 이상 13살 미만이면 어린이 요금 5000원
- 13살 이상 19살 미만이면 청소년 요금 10000원
- 19살 이상 65살 미만이면 성인 요금 15000원
- 65살 이상이면 경로 요금 8000원

````python
age = int(input("나이를 입력하시오."))  #상대방이 나이를 입력할 수 있도록 보완
if age >= 65:  # 이상과 미만에 대해 이해하기 부호 판단 잘하기 >= 등
    print ("경로 요금 8000원")
elif age >= 19:
    print ("성인 요금 15000원")
elif age >= 13:
    print ("청소년 요금 10000원")
elif age >= 3:
    print ("어린이 요금 5000원")
else:
    print ("무료 입장")
````

### 중첩 대신 and

#### 예시문제 : 놀이공원 탑승 확인 프로그램

#### 조건

- 키가 120cm 이상이고, 나이가 8살 이상이면 탑승 가능
- 둘 중 하나라도 만족하지 않으면 탑승 불가

````python
height = int(input("키를 입력하시오."))
age = int(input("나이를 입력하시오"))

if height >= 120 and age >= 8: # 코드 다 만들어 놓고 if 빼먹음 (주의)
    print ("탑승 가능합니다.")
else:
    print("탑승 불가합니다.")
````
